# Model Design Strategy

**Objective:**
* Identify modeling techniques based on real-world data characteristics.
* Prepare testing scenarios and evaluation metrics.
* Implement Decision Tree, Linear Regression, Random Forest, and K-Means models using built-in standard datasets.

```python
# Import necessary global libraries
import pandas as pd
import numpy as np
from sklearn import datasets
from sklearn.model_selection import cross_validate, train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    accuracy_score, precision_score, recall_score, f1_score, confusion_matrix,
    silhouette_score
)
import warnings
warnings.filterwarnings('ignore')

print('Global libraries imported successfully.')
```

## 1. Medical Diagnosis Prediction (Classification)
**Technique:** Decision Tree

**Evaluation:** 5-Fold Cross-Validation (Accuracy, Precision, Recall, F1-Score)

### Step 1: Data Preparation
Load the standard Breast Cancer dataset for binary classification (Malignant vs. Benign).

```python
# Load built-in breast cancer dataset
cancer_data = datasets.load_breast_cancer(as_frame=True)
df_cancer = cancer_data.frame

print("Dataset shape:", df_cancer.shape)
df_cancer.head()
```

### Step 2: Data Preprocessing
Separate the features from the target label. No encoding is needed as the target is already numerical (0 and 1).

```python
# Define feature matrix (X) and target vector (y)
X_cancer = cancer_data.data
y_cancer = cancer_data.target

print("Features mapped. Target names:", cancer_data.target_names)
```

### Step 3: Model Training & Evaluation
Instantiate the Decision Tree model and execute a 5-fold cross-validation to gather robust performance metrics.

```python
# Initialize the classification model
dt_model = DecisionTreeClassifier(random_state=42, max_depth=5)

# Execute 5-fold cross-validation computing multiple metrics simultaneously
scoring_metrics = ['accuracy', 'precision', 'recall', 'f1']
cv_results = cross_validate(dt_model, X_cancer, y_cancer, cv=5, scoring=scoring_metrics)

# Extract and print the mean of the evaluation scores
print('--- Decision Tree (5-Fold CV) Evaluation ---')
print(f"Mean Accuracy : {np.mean(cv_results['test_accuracy']):.4f}")
print(f"Mean Precision: {np.mean(cv_results['test_precision']):.4f}")
print(f"Mean Recall   : {np.mean(cv_results['test_recall']):.4f}")
print(f"Mean F1-Score : {np.mean(cv_results['test_f1']):.4f}")
```

## 2. California House Price Prediction (Regression)
**Technique:** Linear Regression

**Evaluation:** R-Squared, MAE, RMSE

### Step 1: Data Preparation
Load the California Housing dataset to predict median house values based on location and demographic data.

```python
# Load built-in California housing dataset
housing_data = datasets.fetch_california_housing(as_frame=True)
df_housing = housing_data.frame

print("Housing data loaded. Rows:", len(df_housing))
df_housing.head()
```

### Step 2: Data Preprocessing
Isolate the independent variables and the target price, then split the data into training and testing sets.

```python
# Assign feature matrix and target vector
X_house = housing_data.data
y_house = housing_data.target

# Split data (80% training, 20% testing)
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

print("Training shape:", X_train_h.shape)
print("Testing shape:", X_test_h.shape)
```

### Step 3: Model Training & Evaluation
Fit the Linear Regression model to the training data and evaluate using the test set.

```python
# Train the linear regression model
lr_model = LinearRegression()
lr_model.fit(X_train_h, y_train_h)

# Generate predictions based on the test features
y_pred_house = lr_model.predict(X_test_h)

# Calculate regression evaluation metrics
r2 = r2_score(y_test_h, y_pred_house)
mae = mean_absolute_error(y_test_h, y_pred_house)
rmse = np.sqrt(mean_squared_error(y_test_h, y_pred_house))

print('--- Linear Regression Evaluation ---')
print(f'R-Squared : {r2:.4f}')
print(f'MAE       : {mae:.4f}')
print(f'RMSE      : {rmse:.4f}')
```

## 3. Wine Quality Categorization (Classification)
**Technique:** Random Forest

**Evaluation:** Confusion Matrix, Precision, Recall, F1-Score (Weighted)

### Step 1: Data Preparation
Load the Wine dataset which categorizes wines into three distinct classes based on chemical analysis.

```python
# Load built-in wine dataset
wine_data = datasets.load_wine(as_frame=True)
df_wine = wine_data.frame

print("Wine dataset ready. Rows:", len(df_wine))
df_wine.head()
```

### Step 2: Data Preprocessing
Split the dataset into training and testing segments to evaluate the ensemble classifier properly.

```python
# Define feature matrix and target vector
X_wine = wine_data.data
y_wine = wine_data.target

# Split data (80% training, 20% testing)
X_train_w, X_test_w, y_train_w, y_test_w = train_test_split(X_wine, y_wine, test_size=0.2, random_state=42)

print("Target Classes:", wine_data.target_names)
```

### Step 3: Model Training & Evaluation
Utilize the Random Forest algorithm to classify the wine types. Calculate weighted metrics to account for multi-class targets.

```python
# Initialize and train the Random Forest ensemble model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_w, y_train_w)

# Predict target values
y_pred_wine = rf_model.predict(X_test_w)

# Calculate classification metrics using 'weighted' average for multi-class
cm = confusion_matrix(y_test_w, y_pred_wine)
precision = precision_score(y_test_w, y_pred_wine, average='weighted')
recall = recall_score(y_test_w, y_pred_wine, average='weighted')
f1 = f1_score(y_test_w, y_pred_wine, average='weighted')

print('--- Random Forest Evaluation ---')
print('Confusion Matrix:\n', cm)
print(f'Precision (Weighted) : {precision:.4f}')
print(f'Recall (Weighted)    : {recall:.4f}')
print(f'F1-Score (Weighted)  : {f1:.4f}')
```

## 4. Botanical Species Segmentation (Clustering)
**Technique:** K-Means Clustering

**Evaluation:** Inertia, Silhouette Score

### Step 1: Data Preparation
Load the Iris dataset to perform unsupervised clustering based on sepal and petal dimensions.

```python
# Load built-in Iris dataset
iris_data = datasets.load_iris(as_frame=True)
df_iris = iris_data.frame

# For clustering, we drop the target column to simulate unsupervised learning
X_iris_raw = df_iris.drop(columns=['target'])

print("Iris data loaded for segmentation. Features:", list(X_iris_raw.columns))
```

### Step 2: Data Preprocessing
Scale the numerical data using StandardScaler. This is critical for K-Means to ensure distance calculations are not biased by varying measurement scales.

```python
# Normalize numerical features
scaler = StandardScaler()
X_iris_scaled = scaler.fit_transform(X_iris_raw)

print("Data scaling completed. Matrix shape:", X_iris_scaled.shape)
```

### Step 3: Model Training & Evaluation
Apply the K-Means algorithm (k=3 segments) and evaluate cluster density and separation using Inertia and the Silhouette Score.

```python
# Execute K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_iris['Assigned_Cluster'] = kmeans.fit_predict(X_iris_scaled)

# Retrieve internal evaluation metrics
inertia = kmeans.inertia_
sil_score = silhouette_score(X_iris_scaled, df_iris['Assigned_Cluster'])

print('--- K-Means Clustering Evaluation ---')
print(f'Inertia          : {inertia:.4f}')
print(f'Silhouette Score : {sil_score:.4f}')
print('\nCluster Distribution:')
print(df_iris['Assigned_Cluster'].value_counts())
```
